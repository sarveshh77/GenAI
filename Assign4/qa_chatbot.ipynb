Build a simple question-answering chatbot using a pre-trained language model with an interactive interface.

```python
# Cell 2: Import and Load Model
from transformers import pipeline
import gradio as gr

qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

print("✓ Q&A model loaded successfully")
```

```python
# Cell 3: Create Context Database
contexts = {
    "company": """Our company was founded in 2020. We specialize in AI solutions for businesses. 
                  We have offices in New York, London, and Tokyo with over 500 employees.""",
    "products": """We offer three main products: AI Analytics Platform, ChatBot Builder, 
                   and Predictive Modeling Suite. All products include 24/7 support.""",
    "support": """Customer support is available via email, phone, and live chat. 
                  Premium customers get dedicated account managers."""
}

print("✓ Context database created")
print(f"Available topics: {list(contexts.keys())}")
```

```python
# Cell 4: QA Function
def answer_question(question, context_type="company"):
    context = contexts.get(context_type, contexts["company"])
    result = qa_pipeline(question=question, context=context)
    return f"Answer: {result['answer']}\nConfidence: {result['score']:.2%}"

# Test the function
test_answer = answer_question("When was the company founded?", "company")
print("Test question: When was the company founded?")
print(test_answer)
```

```python
# Cell 5: Gradio Interface
interface = gr.Interface(
    fn=answer_question,
    inputs=[
        gr.Textbox(label="Your Question"),
        gr.Dropdown(choices=list(contexts.keys()), label="Topic")
    ],
    outputs=gr.Textbox(label="Answer"),
    title="Q&A Chatbot",
    description="Ask questions about our company, products, or support!"
)

print("✓ Gradio interface created")
```

```python
# Cell 6: Launch Interface
interface.launch(share=True)

print("✓ Chatbot interface launched!")
```

```python
# Cell 7: Command-Line Version
def chat():
    print("Q&A Chatbot (type 'quit' to exit)")
    print(f"Available topics: {list(contexts.keys())}\n")
    
    while True:
        question = input("\nYour question: ")
        if question.lower() == 'quit':
            break
        
        topic = input(f"Topic ({'/'.join(contexts.keys())}): ") or "company"
        print(answer_question(question, topic))

# Uncomment to run the command-line version
# chat()
```
